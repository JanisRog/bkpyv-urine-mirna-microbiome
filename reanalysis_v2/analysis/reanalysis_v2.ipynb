# Revised BKPyV analysis
Uses integer counts, DESeq2/edgeR and verified accession-level bacterial profiles. Install the Python requirements and R packages described in README first. The local package includes small inputs; the code-only repository requires approved inputs from the authors. Choose a new output folder for each run. This does not complete the outstanding upstream DNA classification.

```python
from pathlib import Path
import subprocess, sys

candidates = [base for parent in [Path.cwd(), *Path.cwd().parents]
              for base in [parent, parent / "reanalysis_v2"]
              if (base / "analysis/run_revision.py").is_file()]
if not candidates:
    raise RuntimeError("Open this notebook from the reanalysis_v2 folder or repository")
package = candidates[0]

def choose(label, default):
    answer = input(f"{label} [{default}]: ").strip()
    path = Path(answer).expanduser() if answer else default
    if not path.exists():
        raise FileNotFoundError(f"{label}: {path}")
    return path.resolve()

profiles = choose("DNA profile folder", package / "inputs/dna_profiles")
mapping = choose("ENA mapping TSV", package / "results/dna_qc/ena_sample_mapping.tsv")
metadata = choose("Sample metadata CSV", package / "verified_data/metadata/bkv_sample_metadata.csv")
recount = choose("Recount folder", package / "inputs/recount")
qc = choose("Small RNA QC CSV", package / "inputs/smallRNA_sample_qc.csv")
rlib = choose("R library folder", package / "Rlib")
output_answer = input(f"New output folder [{package / "rerun_output"}]: ").strip()
output = Path(output_answer).expanduser().resolve() if output_answer else package / "rerun_output"
if output.exists():
    raise FileExistsError(f"Choose a new output folder: {output}")
print("Outputs:", output)
```

```python
subprocess.run([sys.executable, str(package / "analysis/run_revision.py"),
    "--profiles", str(profiles), "--mapping", str(mapping),
    "--metadata", str(metadata), "--recount", str(recount), "--qc", str(qc),
    "--rlib", str(rlib), "--out", str(output)], check=True)
```
