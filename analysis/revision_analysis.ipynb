# BKPyV urinary miRNA and microbiome analysis

Run this notebook from the cloned repository using its Python 3.12 environment. Install `analysis/requirements.txt` and notebook support as described in README.md. The input data require permission and are not included.

Enter the local data-folder path at the prompt, or set `BKPYV_DATA_DIR` before launching. The notebook locates the repository from its working directory and leaves input files unchanged. Clear all outputs before committing the notebook.


```python
from pathlib import Path
import os
import subprocess
import sys

cwd = Path.cwd().resolve()
repo_dir = next(
    (p for p in [cwd, *cwd.parents]
     if (p / "analysis/reanalyze.py").is_file() and (p / "docs/INPUTS.md").is_file()),
    None,
)
if repo_dir is None:
    raise RuntimeError("Open the cloned repository folder in VS Code or launch Jupyter from it.")

configured_path = os.environ.get("BKPYV_DATA_DIR") or input("Path to approved input data folder: ").strip()
if not configured_path:
    raise ValueError("Enter the input data folder path.")
data_dir = Path(configured_path).expanduser().resolve()
if not data_dir.is_dir():
    raise FileNotFoundError(f"Input folder does not exist: {data_dir}")
output_dir = repo_dir / "outputs"
```

```python
subprocess.run(
    [sys.executable, str(repo_dir / "analysis/reanalyze.py"),
     "--data-dir", str(data_dir), "--out-dir", str(output_dir)],
    check=True,
)
print("Analysis complete. Tables and figures are in the repository outputs folder.")
```
